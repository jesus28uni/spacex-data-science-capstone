# Limpieza y preparación de datos
Se crea la variable objetivo `Class` (1 = aterrizaje exitoso, 0 = fallido) y se genera `dataset_part_2.csv`.

```python
import pandas as pd, numpy as np
url='https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/api/dataset_part_1.csv'
try:
    df=pd.read_csv('dataset_part_1.csv')
except FileNotFoundError:
    df=pd.read_csv(url)
df.head()
```

```python
print(df.isnull().sum()/len(df)*100)
print(df.dtypes)
```

```python
print(df['LaunchSite'].value_counts())
print(df['Orbit'].value_counts())
```

```python
landing_outcomes=df['Outcome'].value_counts()
for i,o in enumerate(landing_outcomes.keys()):print(i,o)
```

```python
bad_outcomes=set(landing_outcomes.keys()[[1,3,5,6,7]])
print(bad_outcomes)
```

```python
landing_class=[0 if o in bad_outcomes else 1 for o in df['Outcome']]
df['Class']=landing_class
print(df[['Class']].head(8))
print('Tasa de éxito:',df['Class'].mean())
```

```python
df.to_csv('dataset_part_2.csv',index=False)
```
